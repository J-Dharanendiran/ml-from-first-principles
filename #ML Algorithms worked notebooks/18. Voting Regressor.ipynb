```python
import numpy as np
from sklearn.datasets import load_diabetes
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import VotingRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

```python
# 1. Load Dataset
data = load_diabetes()

X = data.data
y = data.target
```

```python
# 3. Understand the data
print("X shape:", X.shape)
print("y shape:", y.shape)

print("Feature names:")
print(data.feature_names)
```

```python
# 4. Create individual models
lr = LinearRegression()
dt = DecisionTreeRegressor(random_state=42)
svr = make_pipeline(StandardScaler(), SVR())
```

```python
# 5. Put the models into a list
estimators = [('lr',lr),('dt',dt),('svr',svr)]
```

```python
# 6. Evaluate the Model individually
for name, model in estimators:
    scores = cross_val_score(model, X,y,scoring='r2', cv=10)
    print(name, 'Mean r2: ', np.round(np.mean(scores),2))
```

```python
# 7. Create voting Regressor
vr = VotingRegressor(estimators=estimators)
```

```python
# 8. Evaluate Voting Regressor
scores = cross_val_score(vr, X, y, scoring='r2', cv=10)
print('Voting Regressor Mean R2:', np.round(np.mean(scores),3))
```

```python
# 9. Trying with Different Weights
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vr = VotingRegressor(estimators=estimators, weights=[i,j,k])
            scores = cross_val_score(vr, X, y, cv=10, scoring='r2')
            print(f'Weights = [{i},{j},{k}]', 'Mean R2:', np.round(np.mean(scores),3))
```
