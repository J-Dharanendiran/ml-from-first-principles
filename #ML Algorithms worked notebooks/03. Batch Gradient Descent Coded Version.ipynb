```python
from sklearn.datasets import load_diabetes

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=2)
```

```python
lr = LinearRegression()

lr.fit(X_train, y_train)
print(lr.coef_)
print(lr.intercept_)
```

```python
y_pred =  lr.predict(X_test)
```

```python
r2 =  r2_score(y_test,y_pred)
r2
```

##  Now, doing the same with the help of Gradient Descent completely from scratch!!

```python
class BatchGDRegressor:

    def __init__(self, learning_rate=0.01, epochs=100):
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X_train, y_train):

        n_samples, n_features = X_train.shape

        # Initialize parameters
        # β = [β₁, β₂, ..., βₙ]
        self.coef_ = np.ones(n_features)

        # β₀ (intercept)
        self.intercept_ = 0

        for epoch in range(self.epochs):

            # Forward pass
            # ŷ = Xβ + β₀
            y_hat = np.dot(X_train, self.coef_) + self.intercept_

            # Errors
            # e = y - ŷ
            errors = y_train - y_hat

            # Gradients
            # ∂L/∂β₀ = -(2/n) Σ(y - ŷ)
            intercept_der = -2 * np.mean(errors)

            # ∂L/∂β = -(2/n) Xᵀ(y - ŷ)
            coef_der = -2 * np.dot(errors, X_train) / n_samples

            # Parameter updates
            # β₀ := β₀ - α(∂L/∂β₀)
            self.intercept_ -= self.lr * intercept_der

            # β := β - α(∂L/∂β)
            self.coef_ -= self.lr * coef_der

    def predict(self, X_test):

        # ŷ = Xβ + β₀
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
model = BatchGDRegressor(0.5, 100)
```

```python
model.fit(X_train, y_train)

print(f"Slope (m): {model.coef_}")
print(f"Intercept (b): {model.intercept_}")
```

```python
# Now, check the R2 Score
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2 = r2_score(y_test,y_pred)
r2
```
