````python
import pandas as pd

from sklearn.preprocessing import OrdinalEncoder
from sklearn.naive_bayes import CategoricalNB
````

````python
data = {
    "Outlook": [
        "Sunny", "Sunny", "Overcast", "Rain", "Rain",
        "Rain", "Overcast", "Sunny", "Sunny", "Rain",
        "Sunny", "Overcast", "Overcast", "Rain"
    ],

    "Temperature": [
        "Hot", "Hot", "Hot", "Mild", "Cool",
        "Cool", "Cool", "Mild", "Cool", "Mild",
        "Mild", "Mild", "Hot", "Mild"
    ],

    "Humidity": [
        "High", "High", "High", "High", "Normal",
        "Normal", "Normal", "High", "Normal", "Normal",
        "Normal", "High", "Normal", "High"
    ],

    "Wind": [
        "Weak", "Strong", "Weak", "Weak", "Weak",
        "Strong", "Strong", "Weak", "Weak", "Weak",
        "Strong", "Strong", "Weak", "Strong"
    ],

    "PlayTennis": [
        "No", "No", "Yes", "Yes", "Yes",
        "No", "Yes", "No", "Yes", "Yes",
        "Yes", "Yes", "Yes", "No"
    ]
}

df = pd.DataFrame(data)
````

````python
df
````

````python
# Separate the features
X = df.drop('PlayTennis', axis=1)
y = df['PlayTennis']
````

````python
# Encode the categorical features
encoder = OrdinalEncoder()

X_encoded = encoder.fit_transform(X)
````

````python
# Encode the target
y_encoded = y.map({"No": 0, "Yes": 1})
````

````python
# Create the Naive Bayes Model
model = CategoricalNB()
````

````python
# Train the model
model.fit(X_encoded, y_encoded)
````

````python
# New Observation
new_day = pd.DataFrame({
    "Outlook": ["Sunny"],
    "Temperature": ["Cool"],
    "Humidity": ["High"],
    "Wind": ["Strong"]
})
````

````python
# Encode the new observation
new_day_encoded = encoder.transform(new_day)
````

````python
# Predict the new observation
prediction = model.predict(new_day_encoded)

prediction_label = "Yes" if prediction[0] == 1 else "No"

print("Prediction:", prediction_label)
````

````python
# Get the probabilities
probabilities = model.predict_proba(new_day_encoded)

print("\nClasses:", model.classes_)
print("Probabilities:", probabilities)
````

# Recommended Learning Path — Naive Bayes

Don't immediately throw `train_test_split`, pipelines, `GridSearchCV`, and a dozen other pieces of machinery at this. You're trying to understand **Naive Bayes**, not recreate NASA's weather department. 😄

Start with this progression:

> **Play Tennis dataset → Calculate probabilities manually → Implement with sklearn → Compare both results**

## Step-by-Step

### 1. Use the Play Tennis dataset
- Understand the features and target.
- Identify the possible classes: `Yes` and `No`.

### 2. Calculate the probabilities manually
Calculate the prior probabilities:

$$
P(\text{Yes}), \quad P(\text{No})
$$

Calculate the conditional probabilities:

$$
P(x_i \mid \text{Yes}) \quad \text{and} \quad P(x_i \mid \text{No})
$$

### 3. Apply the Naive Bayes formula manually

$$
P(C_k \mid X) \propto P(C_k) \prod_{i=1}^{n} P(x_i \mid C_k)
$$

### 4. Implement the same problem using `CategoricalNB`
