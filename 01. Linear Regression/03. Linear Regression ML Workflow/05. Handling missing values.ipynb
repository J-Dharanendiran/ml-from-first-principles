```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_excel("C:\Dharan\Cheat Sheets\Datasets\Data_Train.xlsx")
df.head()
```

### Step 1: Checking for missing values

```python
df.isnull().sum()
```

### Step 2: Calculating the percentage of missing values 

```python
missing_percentage=(df.isnull().sum()/len(df)) * 100
missing_percentage
```

### Step 3: Missing values in the form of DataFrame

```python
missing_df=pd.DataFrame({
    'Missing Values':df.isnull().sum(),
    'Missing Percentage': (df.isnull().sum()/len(df))
})

missing_df
```

### Step 3: Display only columns containing missing values

```python
missing_df[missing_df["Missing Values"] > 0]
```

### Step 5: Remove missing values

```python
df.dropna(inplace=True)
```

### Step 6: verify

```python
df.isnull().sum()
```

### 📝 Notebook conclusion

> The dataset contains only 2 missing values (Route and Total_Stops), accounting for less than 0.02% of the total records. Since the proportion of missing data is negligible, the affected rows were removed using dropna().

> This approach preserves data quality without significantly impacting the dataset size.
