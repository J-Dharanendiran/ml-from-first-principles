```python
import numpy as np

from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import VotingClassifier
```

```python
# 1. Loading the dataset
iris = load_iris()

X = iris.data
y = iris.target

print(iris.feature_names)
print(iris.target_names)
```

```python
# 2. Converting the problem into binary classification
X = X[:, :2]

y = (y != 0).astype(int)
```

```python
# 3. Creating individual classifiers
clf1 = LogisticRegression(max_iter=1000)
clf2 = RandomForestClassifier(random_state=42)
clf3 = KNeighborsClassifier()
```

```python
# 4. Create the estimators list
estimators = [('lr', clf1),('rf',clf2),('knn',clf3)]
```

```python
# 5. Evaluate individual models
for name, model in estimators:
    scores = cross_val_score(model, X,y, cv=10,scoring='accuracy')
    print(name, np.round(np.mean(scores),2))
```

```python
# 6. Hard Voting
hard_voting = VotingClassifier(estimators = estimators,voting='hard')

scores = cross_val_score(hard_voting, X, y, cv=10, scoring='accuracy')

print('Hard Voting:', np.round(np.mean(scores),2))
```

```python
# 7. Soft Voting
soft_voting = VotingClassifier(estimators=estimators,voting='soft')

scores = cross_val_score(soft_voting, X, y, cv=10, scoring='accuracy')

print('Soft Voting:', np.round(np.mean(scores),2))
```

```python
# 8. Weighted Voting

for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):

            vc = VotingClassifier(estimators= estimators, voting='soft', weights=[i,j,k])
            scores = cross_val_score(vc, X, y, cv=10, scoring='accuracy') 
            print(f'weights = [{i},{j},{k}], Accuracy = {np.round(np.mean(scores),2)}')
```
