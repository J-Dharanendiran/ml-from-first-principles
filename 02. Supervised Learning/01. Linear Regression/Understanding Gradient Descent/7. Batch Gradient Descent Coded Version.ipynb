```python
from sklearn.datasets import load_diabetes

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=2)
```

```python
lr = LinearRegression()

lr.fit(X_train, y_train)
print(lr.coef_)
print(lr.intercept_)
```

```python
y_pred =  lr.predict(X_test)
```

```python
r2 =  r2_score(y_test,y_pred)
r2
```

##  Now, doing the same with the help of Gradient Descent completely from scratch!!

```python
class GDRegressor:

    def __init__(self, learning_rate, epochs):

        self.lr = learning_rate
        self.epochs = epochs
        self.tolerance = 1e-6

    def fit(self, X_train, y_train):
        # Initialize Parameters
        # β = [β₁, β₂, β₃, ..., βₙ]
        self.coef_ = np.ones(X_train.shape[1])

        # β₀ (Intercept)
        self.intercept_ = 0

        self.iteration = 0

        while self.iteration < self.epochs:
            # Forward Pass
            # ŷ = Xβ + β₀
            y_hat = np.dot(X_train, self.coef_) + self.intercept_

            # Compute Gradients
            # ∂L/∂β₀ = -(2/N) Σ(y - ŷ)
            intercept_der = -2 * np.mean(y_train - y_hat)

            # ∂L/∂β = -(2/N) Xᵀ(y - ŷ)
            coef_der = -2 * np.dot((y_train - y_hat), X_train) / X_train.shape[0]

            # Store Old Parameters

            old_intercept = self.intercept_
            old_coef = self.coef_.copy()

            # Gradient Descent Update
            # β₀ := β₀ - α(∂L/∂β₀)
            self.intercept_ -= self.lr * intercept_der

            # β := β - α(∂L/∂β)
            self.coef_ -= self.lr * coef_der

            self.iteration += 1
            
            # Early Stopping

            # Stop if
            # |β(new) - β(old)| < tolerance
            # AND
            # |β₀(new) - β₀(old)| < tolerance

            if (
                abs(self.intercept_ - old_intercept) < self.tolerance
                and
                np.all(np.abs(self.coef_ - old_coef) < self.tolerance)
            ):
                print(f"Converged after {self.iteration} iterations.")
                break

    def predict(self, X_test):

        # ŷ = Xβ + β₀
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
model = GDRegressor(0.001, 100)
```

```python
model.fit(X_train, y_train)

print(f"Training completed in {model.iteration} iterations")
print(f"Slope (m): {model.coef_}")
print(f"Intercept (b): {model.intercept_}")
```

```python
# Now, check the R2 Score
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2 = r2_score(y_test,y_pred)
r2
```
